```python
import pandas as pd
import requests
import time
import json
import os
import sys
import ast



from tqdm import tqdm
sys.path.append(os.path.abspath("../src"))
from harmonization import (
    harmonize_automated_classification,
    harmonize_manual_classification,
)

from visualization import (
    compute_metrics_single_method
)

from api import (
    map_cpdat_to_chemsource,
    get_cpdat_puc_superclasses,
)
```

```python
classified_drug_library_data_path = "../data/drug_library/validation_data_classified_all_3_methods.csv"

harmonized_automated = harmonize_automated_classification(classified_drug_library_data_path)
harmonized_manual = harmonize_manual_classification(classified_drug_library_data_path)

drug_library_text = pd.read_csv(classified_drug_library_data_path)
```

```python
input_names = drug_library_text["name_used"].tolist()
feature_ids = drug_library_text.index.tolist()
output_cpdat_data = []
for i, item in tqdm(enumerate(input_names)):
    cpdat_superclasses = get_cpdat_puc_superclasses(item)
    output_cpdat_data.append({
        "feature_id": feature_ids[i],
        "name_used": item,
        "cpdat_puc_superclasses": cpdat_superclasses
    })

cpdat_df = pd.DataFrame(output_cpdat_data)
```

```python
# cpdat_df.to_csv("../data/drug_library/validation_data_cpdat_puc_superclasses.csv")

cpdat_df = pd.read_csv("../data/drug_library/validation_data_cpdat_puc_superclasses.csv")
cpdat_df = cpdat_df.rename(columns={"feature_id": "FEATURE_ID"})
cpdat_df["cpdat_puc_superclasses"] = cpdat_df["cpdat_puc_superclasses"].apply(lambda x: ast.literal_eval(x) if pd.notnull(x) else None)
```

```python
all_cpdat_superclasses = []
for superclasses in cpdat_df["cpdat_puc_superclasses"].dropna():
    for superclass in superclasses:
        all_cpdat_superclasses.append(superclass)

unique_cpdat_superclasses = set(all_cpdat_superclasses)
print(unique_cpdat_superclasses)
# get value counts
cpdat_superclass_counts = {}
for superclass in all_cpdat_superclasses:
    if superclass not in cpdat_superclass_counts:
        cpdat_superclass_counts[superclass] = 0
    cpdat_superclass_counts[superclass] += 1

print(cpdat_superclass_counts)

# Filter out 'Medical/dental' superclass (observed only 4 times, not enough
# for additional category analysis)

cpdat_df["cpdat_puc_superclasses"] = cpdat_df["cpdat_puc_superclasses"].apply(lambda x: [sc for sc in x if sc != 'Medical/dental'] if x is not None else x)
```

```python
cpdat_df["cpdat_chemsource"] = cpdat_df["cpdat_puc_superclasses"].apply(map_cpdat_to_chemsource)

harmonized_cpdat_subset = cpdat_df[["FEATURE_ID", "cpdat_chemsource"]][cpdat_df["cpdat_chemsource"].notnull()]
harmonized_cpdat_exploded = harmonized_cpdat_subset['cpdat_chemsource'].str.join(',').str.get_dummies(sep=',').astype(bool)
harmonized_cpdat_exploded = harmonized_cpdat_exploded.astype(int)
harmonized_cpdat_subset = pd.concat([harmonized_cpdat_subset[["FEATURE_ID"]], harmonized_cpdat_exploded], axis=1)
```

```python
harmonized_cpdat_subset.to_csv("../data/drug_library/harmonized_cpdat_chemsource_categories.csv")
```
